# Upload model.bim as a Fabric semantic model

**Before running:**
1. Upload `microsoft/model.bim` (from the repo) into this workspace's Lakehouse, under **Files**.
2. Attach this notebook to that same Lakehouse.

`model.bim` already has its Direct Lake `DatabaseQuery` expression pointed at your real
workspace/Lakehouse (baked in by `export_semantic_model.py` via `FABRIC_WORKSPACE_ID`/
`FABRIC_LAKEHOUSE_ID` in `.env`) - this notebook just uploads it via `semantic-link-labs`.

Note: the model's Direct Lake partitions expect a `dbo.customers` / `dbo.orders` table (schema
`dbo`) in this Lakehouse. If your tables live under a different schema name, re-export
`model.bim` with `FABRIC_SCHEMA` set accordingly before uploading it here.

```python
import json

import sempy.fabric as fabric
from sempy_labs import create_semantic_model_from_bim, update_semantic_model_from_bim

with open("/lakehouse/default/Files/model.bim") as f:
    bim = json.load(f)
```

```python
dataset = bim["name"]
workspace_id = fabric.get_workspace_id()

try:
    create_semantic_model_from_bim(dataset=dataset, bim_file=bim, workspace=workspace_id)
except ValueError:
    print(f"'{dataset}' already exists in this workspace - updating instead")
    update_semantic_model_from_bim(dataset=dataset, bim_file=bim, workspace=workspace_id)
```
